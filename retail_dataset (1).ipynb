**Step 21 : Hypothesis Testing**

```python
 # 21a — Pearson correlation (unit_price vs sales_amount)
r, p = stats.pearsonr(df["unit_price"], df["sales_amount"])
print(f"Pearson r = {r:.3f}, p-value = {p:.2e}")
print("Reject H0 -> significant correlation" if p < 0.05 else "Fail to reject H0")

# 21b — t-test (sales_amount: returned vs not returned orders)
returned = df.loc[df["return_flag"] == True, "sales_amount"].dropna()
not_returned = df.loc[df["return_flag"] == False, "sales_amount"].dropna()

shapiro_ret = stats.shapiro(returned.sample(min(500, len(returned)), random_state=1))
levene_stat, levene_p = stats.levene(returned, not_returned)
print(f"Shapiro-Wilk (returned sample) p = {shapiro_ret.pvalue:.4f}")
print(f"Levene's test p = {levene_p:.4f}")

t_stat, t_p = stats.ttest_ind(returned, not_returned, equal_var=(levene_p > 0.05))
print(f"t-statistic = {t_stat:.3f}, p-value = {t_p:.4f}")
print("Reject H0 -> sales amount differs by return status" if t_p < 0.05 else "Fail to reject H0")

```

# Step 22 — Identify Key Metrics / KPIs for Dashboard
Chosen based on the problem statement and the Retail/Sales domain — each is measurable, tied to the problem statement, and drives an actual decision:


1. Total Revenue (sales_amount summed)
2. Total Orders and Delivered Orders
3. Cancellation Rate % (Cancelled ÷ total)
4. Within-Target Shipping % (days_to_ship ≤ 5, among orders that were actually delivered)
5. Average Order Value and Average Discount %
6.Average Rating
7. Return Rate %
8. Revenue by City (identifies top/bottom performing cities)

```python
delivered_mask = df["order_status"] == "Delivered"

# Within-Target Shipping %: days_to_ship is meaningless for orders that never
# actually shipped (Cancelled / Pending never left the warehouse) -- so
# computing "days_to_ship <= 5" over ALL orders silently counts every
# cancellation/pending order as neither on nor off target in a way that
# distorts the rate (48.33% vs. the correct 47.93% once non-delivered orders
# are excluded from the denominator). Define it only over orders that were
# actually delivered.
kpis = {
    "Total Revenue (INR)": df["sales_amount"].sum(),
    "Total Orders": len(df),
    "Delivered Orders": int(delivered_mask.sum()),
    "Cancellation Rate %": (df["order_status"] == "Cancelled").mean() * 100,
    "Return Rate %": df["return_flag"].mean() * 100,
    "Within-Target Shipping % (of delivered orders)": (df.loc[delivered_mask, "days_to_ship"] <= 5).mean() * 100,
    "Average Order Value (delivered orders)": df.loc[delivered_mask, "sales_amount"].mean(),
    "Average Discount %": df["discount_pct"].mean() * 100,
    "Average Rating": df["customer_satisfaction"].mean(),
}
kpi_df = pd.DataFrame.from_dict(kpis, orient="index", columns=["value"]).round(2)
kpi_df
```

```python
df["order_weekday"] = df["order_date"].dt.day_name()

heat_data = df.pivot_table(index="order_weekday", columns="city", values="order_id", aggfunc="count")
weekday_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
heat_data = heat_data.reindex(weekday_order)

fig, ax = plt.subplots(figsize=(13, 5))
sns.heatmap(heat_data, cmap="Blues", annot=True, fmt=".0f", ax=ax)
ax.set_title("Order Volume Heatmap — Weekday x City")
plt.tight_layout()
plt.show()
```

```python
funnel_stages = {
    "Booked": len(df),
    "Not Cancelled": (df["order_status"] != "Cancelled").sum(),
    "Delivered": delivered_mask.sum(),
    "Delivered Within Target (<=5 days)": ((delivered_mask) & (df["days_to_ship"] <= 5)).sum(),
}
funnel_df = pd.DataFrame({"stage": list(funnel_stages.keys()), "count": list(funnel_stages.values())})
funnel_df["pct_of_booked"] = (funnel_df["count"] / funnel_df["count"].iloc[0] * 100).round(1)

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.barh(funnel_df["stage"], funnel_df["count"], color="#3b6fa0")
for i, (c, p) in enumerate(zip(funnel_df["count"], funnel_df["pct_of_booked"])):
    ax.text(c, i, f"  {c} ({p}%)", va="center")
ax.invert_yaxis()
ax.set_title("Booking -> Delivered-On-Target Funnel")
plt.tight_layout()
plt.show()
funnel_df
```

# Conclusion
All 23 steps of the checklist have been applied end-to-end: the raw dataset was inspected (Steps 1–8), cleaned and prepared into retail_clean.csv (Steps 9–17), statistically analyzed across univariate, bivariate, multivariate, and hypothesis-testing lenses (Steps 18–21), and summarized into dashboard-ready KPIs and charts (Steps 22–23).

Key findings to carry into a dashboard:

Profit differs significantly by product category (ANOVA F=35.47, p<0.001) — Electronics and Furniture are the profit engines; Groceries and Books drive volume but contribute little margin.
Unit price is only weakly correlated with sales amount (Pearson r=0.294, p<0.001) — quantity and discount move the final invoice value more than price alone.
Region has no statistically detectable relationship with order outcome (Chi-square p=0.499) — cancellations/returns aren't concentrated in any one region.
Returned orders don't differ significantly in sales amount from kept orders (t-test p=0.491).
Within-target shipping (≤ 5 days) sits at 47.9% among orders that were actually delivered — the 48.3% figure some cuts of this data show is a rounding artifact of including Cancelled/Pending orders (which never shipped) in the denominator.
Fixes applied during review: the Within-Target Shipping KPI was originally computed over all orders, which implicitly treats Cancelled and Pending orders (days_to_ship recorded but the order never actually completed) as part of the shipping-performance population. That's now restricted to Delivered orders only, since a Cancelled order was never "on time" or "late" — it just didn't ship. Average Order Value was similarly recomputed over Delivered orders only, since Cancelled/Returned orders' sales_amount reflects what was billed, not what was ultimately fulfilled.
